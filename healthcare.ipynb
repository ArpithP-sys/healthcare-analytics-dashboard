```python
# data cleaning
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# Fixing name format
df['Name']=df['Name'].str.strip().str.title()
```

```python
df
```

```python
# fixing negative bill amounts
df.loc[df['Billing Amount']<0,'Billing Amount']=df['Billing Amount'].mean()
```

```python
df['Billing Amount'].min()
```

```python
df.describe()
```

```python
# since discharge date and date of admission are in string format, we need to convert them to datetime format
df['Date of Admission']=pd.to_datetime(df['Date of Admission'])
df['Discharge Date']=pd.to_datetime(df['Discharge Date'])
```

```python
df.info()
```

```python
df.dtypes
```

```python
df['Length_of_Stay']=(df['Discharge Date']-df['Date of Admission']).dt.days
```

```python
# final cleaning check
df.isnull().sum()

```

```python
# saving cleaned data to a new csv file
df.to_csv('cleaned_healthcare_dataset.csv',index=False)
```

Exploratory Data Analysis and Visualization

```python
# exploratory data analysis and visualization
import matplotlib.pyplot as plt
# Which medical conditions affects the most patients across the network?
condition_counts=df['Medical Condition'].value_counts()
condition_counts
```

```python
plt.figure(figsize=(8,5))
plt.bar(condition_counts.index, condition_counts.values, color='#7833FE')
plt.xlabel('Medical Condition')
plt.ylabel('Number of Patients')
plt.title('Distribution of Medical Conditions')
plt.xticks(rotation=45) #rotate x-axis labels for better visibility
plt.show()
```

```python
# How is patient age distributed across the network?
average_age=df['Age'].mean()
average_age

plt.Figure(figsize=(8,5))
plt.hist(df['Age'], bins=10, color='#FF5733', edgecolor='black')
plt.axvline(average_age, color='blue', linestyle='dashed', linewidth=1, label=f'Average Age: {average_age:.2f}')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.title('Distribution of Patient Age')
plt.legend()
```

```python
# How has average treatment cost changed over time?
average_cost=df['Billing Amount'].mean()
average_cost
```

```python
df['Admission Month']=df['Date of Admission'].dt.to_period('M')
monthly_avg_cost=df.groupby('Admission Month')['Billing Amount'].mean().plot(kind='line', marker='o', color='#33FF57')
plt.xlabel('Admission Month')
plt.ylabel('Average Billing Amount')
plt.title('Average Treatment Cost Over Time')
plt.xticks(rotation=45)
```

```python
df
```

```python
# is there a relationship between patient age and treatment code?
# 60+ age group
senior_avg_cost=df.loc[df['Age']>=60,'Billing Amount'].mean()
senior_avg_cost

```

```python
sample_df=df.sample(n=1000,random_state=1)#to get a random sample of 1000 rows from the dataset for further analysis
sample_df.head()
```

```python
plt.Figure(figsize=(8,5))
plt.scatter(sample_df['Age'],sample_df['Billing Amount'], color='#FF33A1', alpha=0.4,s=15) #alpha means transparency of the points, so that overlapping points can be seen more clearly
#s is the size of the points in the scatter plot
plt.xlabel('Age')
plt.ylabel('Billing Amount')
plt.title('Relationship between Age and Treatment Cost')
```

```python
# which insurance provider covers the most patients?

insurance_count=df['Insurance Provider'].value_counts()
insurance_count
```

```python
#pie chart for insurance providers
plt.figure(figsize=(8,8))
plt.pie(insurance_count, labels=insurance_count.index, autopct='%1.1f%%',colors=['#FF33A1','#33FFA1','#33A1FF','purple','yellow'])#autopct is to show percentage on the pie chart
plt.title('Distribution of Patients by Insurance Provider')
plt.tight_layout() #tight_layout is to make sure that the pie chart fits within the figure area and doesn't get cut off
plt.show()
```

Ad-Hoc Questions


```python
# Emergency department capacity planning
# scenario:the hospital is planning staffing for emergency cases.They know how many emergency admissions occurred in each month 
df.dtypes
```

```python
emergency_cases=(df.loc[df['Admission Type']=='Emergency'].groupby(df['Admission Month'].dt.month).size())
emergency_cases
```

```python
# identify high cost patients
# hopsital finance team wants to identify patients with high treatment costs to analyze the reasons behind it and take necessary actions.
high_cost_patients=df.loc[df['Billing Amount']>df['Billing Amount'].quantile(0.9)]
high_cost_patients
```

```python
# flagging emergency cases with abnormal results
# hospital management wants to identify eemergency patients whose test results are abnormal so these cases cna be prioritized for review
priority_cases=df.loc[(df['Admission Type']=='Emergency') & (df['Test Results']=='Abnormal')]
priority_cases
```
